```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student-mat.csv", sep=";")

df.head()
```

```python
average_grade = df["G3"].mean()
print("Average Final Grade:", average_grade)
```

```python
students_above_15 = (df["G3"] > 15).sum()

print("Students who scored above 15:", students_above_15)
```

```python
correlation = df["studytime"].corr(df["G3"])

print("Correlation between Study Time and Final Grade:", correlation)
```

```python
print("Female Average:", df[df["sex"] == "F"]["G3"].mean())
print("Male Average:", df[df["sex"] == "M"]["G3"].mean())
```

```python
plt.figure(figsize=(6,5))

plt.hist(df["G3"], bins=10)

plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Grades")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.scatterplot(data=df, x="studytime", y="G3")

plt.xlabel("Study Time")
plt.ylabel("Final Grade (G3)")
plt.title("Study Time vs Final Grade")

plt.show()
```

```python
gender_avg = df.groupby("sex")["G3"].mean()

plt.figure(figsize=(6,5))

gender_avg.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.title("Average Final Grade by Gender")

plt.show()
```
